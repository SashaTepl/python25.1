Библиотека Pandas

```python
import numpy as np
import pandas as pd
```

```python
# Pandas - расширение Numpy (структурированные массивы). Строки и столбцы индексируются метками, а не только числами
# Series, DataFrame, Index
```

```python
## Series

data = pd.Series([0.25,0.5,0.75,1.0])
print(data)
print(type(data))
print(type(data.values))
print(type(data.index))
```

```python
data = pd.Series([0.25,0.5,0.75,1.0])
print(data[0])
print(data[1:3])
```

```python
data = pd.Series([0.25,0.5,0.75,1.0], index=['a', 'b', 'c', 'd'])
print(data)

print(data['a'], data[0])

print(data['b':'d'])

print(type(data.index))
```

```python
data = pd.Series([0.25,0.5,0.75,1.0], index=[1, 10, 7, 'd'])
print(data)

print(data[1])
print(data[10:'d'])
```

```python
population_dict= {'city_1': 1001, 'city_2': 1002, 'city_3': 1003, 'city_4': 1004, 'city_5': 1005}

population = pd.Series(population_dict)
print(population)
print(population['city_4':'city_5'])

#Для создания Series можно использовать
# - Списки python или массивы numpy
# - скалярное значение
# - словари
```

```python
## DataFrame - двумерный массив с явно определенными индексами. Последовательность согласованных по индексам sеries
```

```python
population_dict= {'city_1': 1001, 'city_2': 1002, 'city_3': 1003, 'city_4': 1004, 'city_5': 1005}
area_dict = {'city_1': 9991, 'city_2': 9992, 'city_3': 9993, 'city_4': 9994, 'city_5': 9995}

population = pd.Series(population_dict)
area = pd.Series(area_dict)

# print(population)
# print(area)

states= pd.DataFrame({'population': population, 'area1':area})
print(states)

print(states.values)
print(states.index)
print(states.columns)

print(type(states.values))
print(type(states.index))
print(type(states.columns))

print(states['area1'])

# DataFrame. Способы создания
# -через объекты Series
# - Списки словарей
# - Словари объектов Series
# - двумерный массив numpy
# - структуризированный массив numpy
```

```python
# Index - способ организации ссылки на данные объектов Series и DataFrame. Index-неизменяем, упорядочен, является мультимножеством (могут быть повторяющиеся значения)
```

```python
ind = pd.Index([1,2,3,4,5])
print(ind[1])
print(ind[::2])

# ind[1] = 5 TypeError: Index does not support mutable operations

# Index следует соглашениям объекта set(python)
```

```python
indA = pd.Index([1,2,3,4,5])
indB = pd.Index([2,3,4,5,6])

print(indA.intersection(indB)) #Пересечение
```

```python
# Выборка данных из Series (как словарь)

data = pd.Series([0.25,0.5,0.75,1.0], index=['a', 'b', 'c', 'd'])

print('a' in data)
print('x' in data)

print(data.keys())
print(list(data.items()))

data['a'] =100
print(data)

data['z'] =1000 #Добавили новый элемент
print(data)
```

```python
## как одномерный массив
data = pd.Series([0.25,0.5,0.75,1.0], index=['a', 'b', 'c', 'd'])

print(data['a':'c'])

print(data[0:2])

print(data[(data>=0.5) & (data<1)])

print(data[['a', 'd']])
```

```python
# атрибуты-индексаторы
data = pd.Series([0.25,0.5,0.75,1.0], index=[1, 3, 10, 15])

print(data[1])
print(data.loc[1])
print(data.iloc[1])
```

```python
# Выборка данных из DataFrame (как словарь)

population_dict= {'city_1': 1001, 'city_2': 1002, 'city_3': 1003, 'city_4': 1004, 'city_5': 1005}
area_dict = {'city_1': 9991, 'city_2': 9992, 'city_3': 9993, 'city_4': 9994, 'city_5': 9995}

pop = pd.Series({'city_1': 1001, 'city_2': 1002, 'city_3': 1003, 'city_4': 1004, 'city_5': 1005})
area = pd.Series({'city_1': 9991, 'city_2': 9992, 'city_3': 9993, 'city_4': 9994, 'city_5': 9995}
)
data = pd.DataFrame({'area1':area, 'pop1':pop, 'pop': pop})

print(data)
print(data['area1'])
print(data.area1)

print(data.pop1 is data['pop1'])
print(data.pop is data['pop'])

data['new'] = data['area1']/data['pop1'] #add new
print(data)

#двумерный Numpy-массив
```

```python
data = pd.DataFrame({'area1':area, 'pop1':pop})
print(data)
print(data.values) #2мерный нампи

print(data.T) #transpose

#Доступ к стороке осуществляетсчя через values, дальше массив нампи
print(data['area1'])

print(data.values[0:3])
```

```python
#атрибуты-индексаторы

data = pd.DataFrame({'area1':area, 'pop1':pop, 'pop':pop})

print(data)
print(data.iloc[:3, 1:2])
print(data.loc[:'city_4', 'pop1':'pop'])
print(data.loc[data['pop'] >1002, ['area1','pop']])

data.iloc[0,2] = 9999
print(data)
```

```python
rng = np.random.default_rng()
s = pd.Series(rng.integers(0,12,4))

print(s)
print(np.exp(s))
```

```python
#NaN - not a number
#NA - значения: NaN, null

# Pandas. Два спососба хранения отсутсвующих значний
# ИНдикаторы NaN, None
# null

#None-объект (накладные расходы). Не работает с сум, мин и тд.

vall = np.array([1,2,3])
print(vall.sum())

#TypeError: unsupported operand type(s) for +: 'int' and 'NoneType'
# vall = np.array([1,2,3, None])
# print(vall.sum())

vall = np.array([1,2, np.nan,3])
print(vall.sum())
print(np.nansum(vall))
```

```python
x = pd.Series(range(10), dtype=int)
print(x)

x[0]=None
x[1]=np.nan

print(x) # Число заменили числом

x1 = pd.Series(['a', 'b', 'c'])
print(x1)

x1[0] = None
x1[1] = np.nan

print(x1) # Объект заменили объектом
```

```python
#
x2 = pd.Series([1,2,3, np.nan, None, pd.NA])
print(x2)

x3 = pd.Series([1,2,3, np.nan, None, pd.NA], dtype='Int32')
print(x3)

print(x3.isnull())
print(x3[x3.isnull()])
print(x3[x3.notnull()])

print(x3.dropna())
```

```python
df= pd.DataFrame([[1,2,3, np.nan, None, pd.NA], [1,2,3,None,5,6], [1, np.nan, 3,None, np.nan, 6]])

print(df)
print(df.dropna())
print('                         ')
print(df.dropna(axis=0))
print('                         ')
print(df.dropna(axis=1))
```

```python
#how

# -all- все значения NA
print(df.dropna(axis=1, how='all'))

# -any- хотя бы одно значение
print(df.dropna(axis=1, how='any'))

# - thresh = x, остается, если присутсвует МИНИМУМ x НЕПУСТЫХ значений
print(df.dropna(axis=1, thresh=2))
```
