```python
import numpy as np
import pandas as pd
```

```python
# 2. Привести различные способы создания объектов типа DataFrame
# DataFrame. Способы создания
# - через объекты Series
# - списки словарей
# - словари объектов Series
# - двумерный массив NumPy
# - структурированный массив Numpy

# - через объекты Series

population_dict = {'city_1': 4000, 'city_2':4001, 'city_3':4002}
population = pd.Series(population_dict)
temp_dict= {'city_1':20, 'city_2':25, 'city_3':30}
temp=pd.Series(temp_dict)
states= pd.DataFrame({'population': population, 'temp':temp})
print(states)

# - списки словарей
data=[{'city_1': 4000, 'city_2':4001, 'city_3':4002}, {'city_1':20, 'city_2':25, 'city_3':30}]
states=pd.DataFrame(data)
print(states)

# - словари объектов Series
full_dict={'pop': pd.Series([1000, 1001, 1002]), 'temperature': pd.Series([20, 25,30])}
states=pd.DataFrame(full_dict)
print(states)

# - двумерный массив NumPy

data=np.random.sample((2, 3))
states=pd.DataFrame(data)
print(states)

# - структурированный массив Numpy

struct = np.array([(1, 'First', 1000),(2, 'Second', 1001), (3, 'Third', 1002)])
states=pd.DataFrame(struct)
print(states)
```

```python
# 3. Объедините два объекта Series с неодинаковыми множествами ключей (индексов) так, чтобы вместо NaN было установлено значение 1

population_dict= {'city_1': 1001, 'city_2': 1002, 'city_3': 1003, 'city_4': 1004, 'city_5': 1005}
area_dict = {'city_1': 9991, 'city_2': 9992, 'city_3': 9993, 'city_4': 9994}
population = pd.Series(population_dict)
area = pd.Series(area_dict)

print(population+area)

print((population+area).fillna(1))
```

```python
# 4. Переписать пример с транслирование для DataFrame так, чтобы вычитание происходило по СТОЛБЦАМ
rng = np.random.default_rng()
A = rng.integers(0, 10,(3,4))

df = pd.DataFrame(A, columns=['a', 'b', 'c', 'd'])
print(df)
print(df.T.iloc[0])
print((df.T-df.T.iloc[0]).T)
```

```python
# 5. На примере объектов DataFrame продемонстрируйте использование методов ffill() и bfill()

data = {'First':[1000, 1001, 1002, None], 'Second':[None, 2001, 2002, 2003], 'Third':[None, 3002, None, None]}
df = pd.DataFrame(data)
print(df)

# Заполнение пропущенных значений методом ffill() -- заполняет пропущенные значения, двигаясь вперед по данным
print(df.ffill())

# Заполнение пропущенных значений методом bfill() -- заполняет пропущенные значения, двигаясь назад по данным
print(df.bfill())
```
