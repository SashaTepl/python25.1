**Массивы**

```python
np.random.seed(1) #с начальным значением 1, случайные числа всегда в одном и том же порядке

x1 = np.random.randint(10, size =3)
print(x1)

x2 = np.random.randint(10, size =(3,2))
print(x2)

x3 = np.random.randint(10, size =(3,2,2))
print(x3)
```

```python
#Свойства массивов

#ndim число размерностей
#shape размер каждой размерности
#size общий размер массива(число элементов)

print(x1.ndim, x1.shape, x1.size)
print(x2.ndim, x2.shape, x2.size)
print(x3.ndim, x3.shape, x3.size)
```

```python
#ДОступ к элементам массива. Индексы (с 0)

a = np.array([1, 2, 3, 4, 5])

print(a[0])
print(a[-2])
a[1] = 20
print(a)

b = np.array([[1,2], [3,4]])
print(b[0,0])
print(b[-1,-1])
b[1, 0] = 100
print(b)
```

```python
a = np.array([1, 2, 3, 4])
b = np.array([1., 2, 3, 4])

print(a)
print(b)

a[0] = 10.9 #Приводится к int, тип в массиве больше не меняется
print(a)
```

```python
# Срез [start:finish:step] [0:shape:1], finish не учитывается

a = np.array([1,2,3,4,5,6])
print(a[0:3:1])
print(a[:3])
print(a[3:])
print(a[1:5])
print(a[1:-1])
print(a[1::2])

print(a[::-1])
```

```python
#Срезы не копии, а части массива!
a = np.array([1,2,3,4,5,6])

b = a[:3]

b[0] = 100

print(a)
```

```python
a = np.arange(1,13)
print(a)
print(a.reshape(2,6)) #Размеры соответсвуют

# x = np.array([1,2,3])
# y = np.array([4,5])
# z = np.array([6])

# print(np.concatenate([x,y,z])) #Объединение

x = np.array([1,2,3])
y = np.array([4,5,6])

r1 = np.vstack([x,y]) #vertical
print(r1)

print(np.hstack([r1,r1])) #horizontal
```

Вычисления с массивами

```python
#Векторизированная операция - независимо  к каждому элементу массива

x = np.arange(10)
print(x)

print(x*2+1)

#Универсальные функции

print(np.add(np.multiply(x,2), 1))

# - - / // ** %
#np. abs, sin/cos/tan, exp, log

x = np.arange(5)

y = np.zeros(10)
print(np.multiply(x, 10, out=y[::2]))
print(y)
```

```python
x = np.arange(1,5)
print(x)

print(np.add.reduce(x)) #СВернули массив к одному значению через сумму

print(np.add.accumulate(x))
```

```python
#векторные произведения
x = np.arange(1,10)
print(np.add.outer(x,x))
print(np.multiply.outer(x,x))
```
